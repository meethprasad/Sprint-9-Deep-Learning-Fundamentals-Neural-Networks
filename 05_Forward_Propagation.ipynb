## 1. What is Forward Propagation?

Forward propagation is the process of passing input data through a neural network from the input layer to the output layer to produce a prediction.

**Flow:**

Input → Linear Transformation → Activation → Output → Prediction


## 2. Main Parts of a Neural Network

### Input Layer
Receives the input features.

### Hidden Layer
Combines inputs using weights and biases and applies an activation function.

### Output Layer
Produces the final network output.

### Weights
Weights control how strongly each input contributes to a neuron.

### Bias
Bias is an additional value added to the weighted sum.

### Activation
An activation function introduces non-linearity.

Here we use ReLU:

ReLU(x) = max(0, x)


## 3. Small Neural Network

We use:

**2 Input Neurons → 2 Hidden Neurons → 1 Output Neuron**

Input:
x = [1, 2]

Hidden layer weights:
W1 = [[0.5, 0.2], [0.3, 0.4]]

Hidden layer bias:
b1 = [0.1, 0.2]

Output layer weights:
W2 = [[0.6], [0.5]]

Output layer bias:
b2 = [0.1]


## 4. Step 1 – Input Layer

The input is:

x = [1, 2]

The two input values are passed to the hidden layer.

## 5. Step 2 – Hidden Layer Linear Transformation

Formula:

z1 = xW1 + b1

For the calculations below, W1 is arranged so that each row represents a hidden neuron.

Hidden neuron 1:

z1[0] = (1 × 0.5) + (2 × 0.3) + 0.1
      = 0.5 + 0.6 + 0.1
      = 1.2

Hidden neuron 2:

z1[1] = (1 × 0.2) + (2 × 0.4) + 0.2
      = 0.2 + 0.8 + 0.2
      = 1.2

Therefore:

z1 = [1.2, 1.2]

## 6. Step 3 – Hidden Layer Activation

We apply ReLU activation.

ReLU(x) = max(0, x)

Therefore:

ReLU([1.2, 1.2]) = [1.2, 1.2]

The activated hidden values are passed to the output layer.

## 7. Step 4 – Output Layer

Formula:

z2 = hW2 + b2

Calculation:

z2 = (1.2 × 0.6) + (1.2 × 0.5) + 0.1

z2 = 0.72 + 0.60 + 0.10

z2 = 1.42

For this simple regression-style example, the output itself is the prediction.

**Prediction = 1.42**

## 8. Manual Calculation Using NumPy

```python
import numpy as np

# Input
x = np.array([1.0, 2.0])

# Hidden layer parameters
# Each row represents one hidden neuron
W1 = np.array([
    [0.5, 0.3],
    [0.2, 0.4]
])

b1 = np.array([0.1, 0.2])

# Output layer parameters
W2 = np.array([
    [0.6],
    [0.5]
])

b2 = np.array([0.1])

print('Input:', x)
print('W1:\n', W1)
print('b1:', b1)
print('W2:\n', W2)
print('b2:', b2)
```

```python
# Step 1: Hidden layer linear transformation
z1 = x @ W1.T + b1

print('Hidden layer before activation:', z1)
```

```python
# Step 2: ReLU activation
h = np.maximum(0, z1)

print('Hidden layer after ReLU:', h)
```

```python
# Step 3: Output layer
z2 = h @ W2 + b2

prediction = z2

print('Output:', z2)
print('Prediction:', prediction[0])
```

## 9. Complete NumPy Forward Pass

The complete process is:

x → xW1 + b1 → ReLU → hW2 + b2 → Prediction

```python
def forward_numpy(x, W1, b1, W2, b2):
    z1 = x @ W1.T + b1
    h = np.maximum(0, z1)
    z2 = h @ W2 + b2
    return z1, h, z2

z1, h, prediction = forward_numpy(x, W1, b1, W2, b2)

print('z1:', z1)
print('Hidden activation:', h)
print('Prediction:', prediction[0])
```

## 10. Same Network Using PyTorch

Now we implement exactly the same network using PyTorch.

```python
import torch

# Input
x_t = torch.tensor([[1.0, 2.0]])

# Hidden layer parameters
W1_t = torch.tensor([
    [0.5, 0.3],
    [0.2, 0.4]
])

b1_t = torch.tensor([0.1, 0.2])

# Output layer parameters
W2_t = torch.tensor([
    [0.6],
    [0.5]
])

b2_t = torch.tensor([0.1])

# Forward propagation
z1_t = x_t @ W1_t.T + b1_t
h_t = torch.relu(z1_t)
z2_t = h_t @ W2_t + b2_t

print('Hidden layer before activation:', z1_t)
print('Hidden layer after ReLU:', h_t)
print('Prediction:', z2_t.item())
```

## 11. Verify NumPy and PyTorch Results

Both implementations should produce the same prediction: **1.42**.

```python
numpy_prediction = float(prediction[0])
pytorch_prediction = z2_t.item()

print('NumPy prediction:', numpy_prediction)
print('PyTorch prediction:', pytorch_prediction)
print('Same result:', np.isclose(numpy_prediction, pytorch_prediction))
```

## 12. How Does an Input Finally Become a Prediction?

1. The input layer receives feature values.
2. Each hidden neuron calculates a weighted sum.
3. A bias is added to the weighted sum.
4. An activation function such as ReLU transforms the value.
5. The activated hidden values are passed to the output layer.
6. The output layer performs another weighted sum and adds a bias.
7. The final output becomes the prediction.

**Simple idea:**

Input → Weights + Bias → Activation → Weights + Bias → Prediction

During training, the neural network learns better weights and biases so that predictions become closer to the target values.

## Viva Points

- Forward propagation means moving data from input to output.
- Weights determine the contribution of inputs.
- Bias shifts the neuron's output.
- Linear transformation is z = xW + b.
- Activation introduces non-linearity.
- ReLU is max(0, x).
- Hidden layers learn intermediate representations.
- The output layer produces the final output.
- A prediction is the result of the forward pass.
