## 1. What is Deep Learning?

Deep Learning is a part of Machine Learning that uses neural networks with multiple layers to learn patterns from data.

**Analogy:** Learning to recognize a dog by gradually noticing simple features such as edges, then shapes, then complete objects.

**Demonstration:** A simple neuron calculation is one of the basic building blocks of a neural network.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn

x = np.array([2.0, 3.0])
w = np.array([0.5, 0.8])
b = 0.2

z = np.dot(x, w) + b
print("Weighted sum:", z)
```

**Result:** Each input is multiplied by a weight, the weighted values are added, and the bias is added. This is the basic computation performed before activation in a neuron.

## 4. Why Neural Networks are Required

Many real-world relationships are nonlinear and complex. Neural networks combine many simple neurons and nonlinear activation functions to learn such patterns.

**Analogy:** One worker can perform a simple task; a team working in stages can solve a more complicated task.


```python
from sklearn.neural_network import MLPRegressor

X = np.linspace(-3, 3, 100).reshape(-1, 1)
y = np.sin(X[:, 0])

mlp = MLPRegressor(
    hidden_layer_sizes=(20,),
    activation="relu",
    max_iter=2000,
    random_state=42
)

mlp.fit(X, y)
pred = mlp.predict(X)

plt.figure(figsize=(8, 5))
plt.scatter(X, y, label="Actual")
plt.plot(X, pred, label="Neural network")
plt.xlabel("X")
plt.ylabel("y")
plt.title("Neural Network Learning a Nonlinear Pattern")
plt.legend()
plt.show()
```

**Result:** The neural network learns an approximation of the nonlinear sine relationship. Hidden neurons and nonlinear activation allow it to model patterns that a simple straight line cannot.

## 6. Supervised Deep Learning

Supervised Deep Learning learns from inputs paired with known target labels.

**Example:** Image → `"cat"`

**Analogy:** A teacher gives examples together with correct answers.

**Demonstration:**


```python
X_supervised = np.array([
    [1, 20],
    [2, 25],
    [3, 30],
    [4, 35],
    [5, 40]
])

y_supervised = np.array([0, 0, 0, 1, 1])

print("Inputs:\n", X_supervised)
print("Labels:", y_supervised)
```

**Result:** Every input has a known target. A supervised neural network learns the relationship between inputs and targets.


## 7. Unsupervised Deep Learning

Unsupervised learning uses data without target labels and attempts to discover structure or useful representations.

Examples include clustering, dimensionality reduction and neural-network-based autoencoders.

**Analogy:** Instead of giving categories to a student, ask the student to group similar objects.


```python
from sklearn.cluster import KMeans

X_unsupervised = np.array([
    [1, 1], [1.5, 1.2], [2, 1.8],
    [8, 8], [8.5, 8.2], [9, 9]
])

kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_unsupervised)

print("Cluster assignments:", clusters)
```

**Result:** Similar points are grouped together without providing class labels. In Deep Learning, neural networks can also learn useful representations without labeled targets.

## 9. Neuron

A neuron receives inputs, multiplies them by weights, adds a bias, and applies an activation function.

**Analogy:** A decision maker considers several pieces of evidence, gives each one a different importance, and produces an output.


```python
x = np.array([2.0, 3.0])
w = np.array([0.5, 0.8])
b = 0.2

z = np.dot(x, w) + b
output = max(0, z)  # ReLU

print("Weighted sum:", z)
print("Neuron output:", output)
```

**Result:** The weighted sum combines the inputs. ReLU keeps positive values and changes negative values to zero.


## 10. Input

Input is the information supplied to the neural network.

Examples: image pixels, audio measurements, customer features, or numerical representations of text.

**Analogy:** Input is the information entering a system.


```python
input_data = np.array([
    [25, 50000],
    [30, 65000],
    [35, 80000]
])

print(input_data)
print("Shape:", input_data.shape)
```

**Result:** There are 3 samples and 2 input features, so the shape is `(3, 2)`.


## 11. Weight

A weight controls how strongly an input influences a neuron. Weights are learned during training.

**Analogy:** A weight is the importance assigned to a clue.


```python
x = np.array([2.0, 3.0])

for w in [np.array([0.1, 0.1]), np.array([1.0, 1.0])]:
    print("Weights:", w, "Weighted sum:", np.dot(x, w))
```

**Result:** Changing weights changes how strongly the inputs contribute to the neuron's output.


## 12. Bias

Bias is an additional learnable value added to the weighted sum.

**Analogy:** Bias is an adjustable starting point for a decision.


```python
x = np.array([2.0, 3.0])
w = np.array([0.5, 0.8])

for b in [0.0, 2.0, -2.0]:
    print("Bias:", b, "Output before activation:", np.dot(x, w) + b)
```

**Result:** Changing the bias shifts the neuron's output without changing the inputs or weights.


## 13. Activation

An activation function transforms the neuron's weighted sum. It introduces nonlinearity so neural networks can learn complex relationships.

Common activations: ReLU, Sigmoid, Tanh and Softmax.

**Analogy:** Activation is a rule determining how strongly a neuron responds.


```python
values = np.array([-2, -1, 0, 1, 2])
relu = np.maximum(0, values)

print("Input:", values)
print("ReLU:", relu)
```

**Result:** ReLU converts negative values to zero and keeps positive values unchanged.


## 14. Layer

A layer is a group of neurons operating at the same stage.

**Analogy:** A layer is like a department containing several workers performing related processing.


```python
layer_input = np.array([1.0, 2.0, 3.0])

weights = np.array([
    [0.2, 0.4, 0.1],
    [0.5, 0.3, 0.2],
    [0.1, 0.6, 0.7]
])

bias = np.array([0.1, 0.2, 0.3])
layer_output = np.dot(layer_input, weights.T) + bias

print("Layer output:", layer_output)
```

**Result:** Three neurons produce three outputs. A layer performs several neuron calculations in parallel.


## 15. Input Layer

The input layer receives the features supplied to the network.

**Analogy:** It is the entrance through which information enters a building.


```python
input_features = 4
print("Number of input features:", input_features)
print("Each sample provides 4 input values.")
```

**Result:** The input layer must match the number of features supplied for each sample.


## 16. Hidden Layer

A hidden layer lies between the input and output layers. Hidden layers transform inputs into useful intermediate representations.

**Analogy:** Hidden layers are processing departments between the entrance and final decision.


```python
from sklearn.neural_network import MLPClassifier

demo_model = MLPClassifier(
    hidden_layer_sizes=(8, 4),
    max_iter=1000,
    random_state=42
)

print("Hidden layer sizes:", demo_model.hidden_layer_sizes)
```

**Result:** This example has two hidden layers: one with 8 neurons and another with 4 neurons.

## 18. Parameters

Parameters are values learned from training data. The main neural-network parameters are weights and biases.

**Analogy:** Parameters are adjustable settings learned from experience.


```python
torch.manual_seed(42)

linear_layer = nn.Linear(3, 2)

print("Weight shape:", linear_layer.weight.shape)
print("Bias shape:", linear_layer.bias.shape)
print("Total parameters:", sum(p.numel() for p in linear_layer.parameters()))
```

**Result:** A linear layer with 3 inputs and 2 outputs has 6 weights and 2 biases, for **8 trainable parameters**.


## 19. Hyperparameters

Examples:

- Number of layers
- Number of neurons
- Batch size
- Number of epochs
- Optimizer
- Dropout rate

**Analogy:** Hyperparameters are like rules selected by a coach before training.


```python
hyperparameters = {
    "learning_rate": 0.001,
    "hidden_layers": 2,
    "neurons_per_layer": 64,
    "batch_size": 32,
    "epochs": 20
}

pd.DataFrame(list(hyperparameters.items()), columns=["Hyperparameter", "Example Value"])
```

**Result:** These values control training/model configuration. They are different from weights and biases learned from data.


## 20. Tensor

A tensor is a multi-dimensional numerical data structure. PyTorch uses tensors as its main data structure.

Examples:
- Scalar → 0D
- Vector → 1D
- Matrix → 2D
- Image batch → commonly 4D

**Analogy:** A tensor is a flexible container that organizes numbers in multiple dimensions.


```python
tensor_0d = torch.tensor(5.0)
tensor_1d = torch.tensor([1.0, 2.0, 3.0])
tensor_2d = torch.tensor([[1.0, 2.0], [3.0, 4.0]])

print("0D shape:", tensor_0d.shape)
print("1D shape:", tensor_1d.shape)
print("2D shape:", tensor_2d.shape)
```

**Result:** The shape tells us how numbers are organized. PyTorch tensors are designed for efficient numerical and neural-network computation.


## 21. Model

A model is the complete mathematical structure used to transform inputs into predictions.

**Analogy:** A model is like a complete machine made from multiple components.


```python
class SimpleNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(2, 4)
        self.activation = nn.ReLU()
        self.output = nn.Linear(4, 1)

    def forward(self, x):
        x = self.layer1(x)
        x = self.activation(x)
        return self.output(x)

model = SimpleNetwork()

print(model)
print("Total parameters:", sum(p.numel() for p in model.parameters()))
```

**Result:** The model contains an input-to-hidden layer, a ReLU activation, and an output layer. Its weights and biases are trainable parameters.


## 22. Complete Neural Network Flow

**Input → Weighted Sum → Bias → Activation → Hidden Layers → Output**

During training:
1. Input passes through the model.
2. The model produces predictions.
3. Loss measures prediction error.
4. Gradients are calculated.
5. Parameters are updated.
6. The process repeats.

**Analogy:** Like a worker receiving information, processing it through several departments, checking the answer, and improving the process repeatedly.


```python
sample_input = torch.tensor([[2.0, 3.0]])

with torch.no_grad():
    prediction = model(sample_input)

print("Input:", sample_input)
print("Model output:", prediction)
```

**Result:** The input passes through the network and produces an output. Since this model has not been trained for a particular task, the output is not expected to be accurate yet.
